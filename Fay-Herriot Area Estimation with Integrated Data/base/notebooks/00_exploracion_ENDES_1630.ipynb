```python
rech23_rec = rech23.copy()

# =========================
# 1. AGUA
# =========================

# HV201:
# 11 Dentro de la vivienda
# 12 Fuera de la vivienda pero dentro del edificio
# 13 Pilón / grifo público
# 21 Pozo en vivienda/patio/lote
# 22 Pozo público
# 41 Manantial
# 43 Río/acequia/laguna
# 51 Agua de lluvia
# 61 Camión cisterna
# 71 Agua embotellada
# 96 Otro

rech23_rec["agua_red"] = rech23_rec["HV201"].isin([11, 12, 13]).astype(int)
rech23_rec["agua_pozo"] = rech23_rec["HV201"].isin([21, 22]).astype(int)
rech23_rec["agua_superficial"] = rech23_rec["HV201"].isin([41, 43]).astype(int)
rech23_rec["agua_lluvia"] = (rech23_rec["HV201"] == 51).astype(int)
rech23_rec["agua_cisterna"] = (rech23_rec["HV201"] == 61).astype(int)
rech23_rec["agua_embotellada"] = (rech23_rec["HV201"] == 71).astype(int)
rech23_rec["agua_otra"] = (rech23_rec["HV201"] == 96).astype(int)

# Tiempo para recoger agua
# 996 = en las instalaciones
# 998 = no sabe
rech23_rec["tiempo_agua"] = rech23_rec["HV204"].replace({996: 0, 998: np.nan})

# =========================
# 2. SANEAMIENTO
# =========================

# HV205:
# 11 Dentro de la vivienda
# 12 Fuera de la vivienda pero dentro del edificio
# 21 Letrina mejorada ventilada
# 22 Pozo séptico
# 23 Letrina (pozo ciego o negro)
# 24 Letrina mejorada colgada/flotante
# 31 Río, acequia o canal
# 32 Sin servicio (matorral/campo)
# 96 Otro

rech23_rec["saneamiento_mejorado"] = rech23_rec["HV205"].isin([11, 12, 21, 22, 24]).astype(int)
rech23_rec["saneamiento_precario"] = rech23_rec["HV205"].isin([23, 31, 32, 96]).astype(int)
rech23_rec["sin_servicio_higienico"] = (rech23_rec["HV205"] == 32).astype(int)

# =========================
# 3. SERVICIOS DEL HOGAR
# =========================

rech23_rec["tiene_electricidad"] = (rech23_rec["HV206"] == 1).astype(int)

# =========================
# 4. MATERIALES DE LA VIVIENDA
# =========================

# HV213 piso:
# 11 tierra/arena
# 21 madera
# 31 parquet
# 32 láminas/vinílicos
# 33 losetas/terrazos
# 34 cemento/ladrillo
# 96 otro

rech23_rec["piso_precario"] = rech23_rec["HV213"].isin([11, 21]).astype(int)
rech23_rec["piso_no_precario"] = rech23_rec["HV213"].isin([31, 32, 33, 34]).astype(int)

# HV214 paredes
rech23_rec["pared_precaria"] = rech23_rec["HV214"].isin([11, 12, 13, 21, 22, 23, 24, 41]).astype(int)
rech23_rec["pared_no_precaria"] = rech23_rec["HV214"].isin([31, 32, 33]).astype(int)

# HV215 techo
rech23_rec["techo_precario"] = rech23_rec["HV215"].isin([11, 12, 21, 22, 41]).astype(int)
rech23_rec["techo_no_precario"] = rech23_rec["HV215"].isin([31, 32, 33, 34]).astype(int)

# Número de habitaciones para dormir
rech23_rec["habitaciones_dormir"] = rech23_rec["HV216"]

# =========================
# 5. BIENES DEL HOGAR
# =========================

# Ojo: según tus valores únicos, HV219 está codificada 1/2, no 0/1
rech23_rec["tiene_telefono"] = (rech23_rec["HV219"] == 1).astype(int)

rech23_rec["tiene_tv"] = (rech23_rec["HV208"] == 1).astype(int)
rech23_rec["tiene_refrigeradora"] = (rech23_rec["HV209"] == 1).astype(int)
rech23_rec["tiene_celular"] = (rech23_rec["HV243A"] == 1).astype(int)

# =========================
# 6. COMBUSTIBLE PARA COCINAR
# =========================

# HV226:
# 1 Electricidad
# 2 GLP
# 3 Gas natural
# 5 Kerosene
# 6 Carbón vegetal
# 7 Carbón mineral
# 8 Leña
# 9 Caña/arbustos secos
# 10 Residuos agrícolas
# 11 Bosta
# 95 No cocina
# 96 Otro

rech23_rec["combustible_limpio"] = rech23_rec["HV226"].isin([1, 2, 3]).astype(int)
rech23_rec["combustible_solido"] = rech23_rec["HV226"].isin([6, 7, 8, 9, 10, 11]).astype(int)
rech23_rec["no_cocina"] = (rech23_rec["HV226"] == 95).astype(int)

# =========================
# 7. TRATAMIENTO DEL AGUA
# =========================

rech23_rec["agua_tratada"] = (
    (rech23_rec["HV237A"] == 1) |
    (rech23_rec["HV237B"] == 1) |
    (rech23_rec["HV237C"] == 1) |
    (rech23_rec["HV237D"] == 1) |
    (rech23_rec["HV237E"] == 1) |
    (rech23_rec["HV237F"] == 1)
).astype(int)

rech23_rec["hervir_agua"] = (rech23_rec["HV237A"] == 1).astype(int)
rech23_rec["clorar_agua"] = (rech23_rec["HV237B"] == 1).astype(int)
rech23_rec["filtrar_pano"] = (rech23_rec["HV237C"] == 1).astype(int)
rech23_rec["agua_filtrada"] = (rech23_rec["HV237D"] == 1).astype(int)
rech23_rec["desinfeccion_solar"] = (rech23_rec["HV237E"] == 1).astype(int)
rech23_rec["dejar_reposar"] = (rech23_rec["HV237F"] == 1).astype(int)

# =========================
# 8. COCINA
# =========================

rech23_rec["cuarto_especial_cocina"] = (rech23_rec["HV242"] == 1).astype(int)

# =========================
# 9. DISPONIBILIDAD DE AGUA
# =========================

rech23_rec["agua_disponible_todo_dia"] = (rech23_rec["SH42"] == 1).astype(int)
```

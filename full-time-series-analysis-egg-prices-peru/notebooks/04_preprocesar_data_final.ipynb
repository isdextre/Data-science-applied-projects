```python
import pandas as pd

df1 = pd.read_csv("tabla_1.csv")
df2 = pd.read_csv("tabla_2.csv")
df3 = pd.read_csv("tabla_3.csv")
```

```python
df1
```

```python
df1['fecha'] = pd.to_datetime(df1['fecha'])
df2['fecha'] = pd.to_datetime(df2['fecha'])
df3['fecha'] = pd.to_datetime(df3['fecha'])
```

```python
# 1. Encontrar máximo y mínimo de cada DataFrame
print("=== MÁXIMOS Y MÍNIMOS POR DATAFRAME ===")
print(f"df1 - Mínimo: {df1['fecha'].min()}, Máximo: {df1['fecha'].max()}")
print(f"df2 - Mínimo: {df2['fecha'].min()}, Máximo: {df2['fecha'].max()}")
print(f"df3 - Mínimo: {df3['fecha'].min()}, Máximo: {df3['fecha'].max()}")

```

```python
print(df1.shape)
print(df2.shape)
print(df3.shape)
```

```python
# Rango total de fechas (desde la mínima global hasta la máxima global)
fecha_min_global = min(df1['fecha'].min(), df2['fecha'].min(), df3['fecha'].min())
fecha_max_global = max(df1['fecha'].max(), df2['fecha'].max(), df3['fecha'].max())


# Crear un rango completo de fechas esperadas
rango_completo = pd.date_range(start=fecha_min_global, end=fecha_max_global, freq='D')

# 2. Calcular fechas faltantes por DataFrame
def calcular_faltantes(df, nombre, rango_completo):
    fechas_presentes = set(df['fecha'].dt.date)
    fechas_esperadas = set(rango_completo.date)
    fechas_faltantes = fechas_esperadas - fechas_presentes
    
    total_esperadas = len(rango_completo)
    cantidad_faltantes = len(fechas_faltantes)
    porcentaje_faltantes = (cantidad_faltantes / total_esperadas) * 100
    
    print(f"\n=== {nombre} ===")
    print(f"Fechas presentes: {len(fechas_presentes)}")
    print(f"Fechas esperadas en el rango: {total_esperadas}")
    print(f"Fechas faltantes: {cantidad_faltantes}")
    print(f"Porcentaje de fechas faltantes: {porcentaje_faltantes:.2f}%")
    
    return fechas_faltantes

# Calcular para cada DataFrame
faltantes_df1 = calcular_faltantes(df1, "DataFrame 1", rango_completo)
faltantes_df2 = calcular_faltantes(df2, "DataFrame 2", rango_completo)
faltantes_df3 = calcular_faltantes(df3, "DataFrame 3", rango_completo)
```

# Nos quedamos con la información necesaria de cada df

```python
df1.drop(columns=['dia', 'mes_txt', 'mes_num', 'anio'], inplace=True)
df2.drop(columns=['dia', 'mes_txt', 'mes_num', 'anio'], inplace=True)
df3.drop(columns=['dia', 'mes_txt', 'mes_num', 'anio'], inplace=True)
# renombramos las columnas de df2
df2.rename(columns={'lima': 'oferta_lima(ton)',
                    'sierra': 'oferta_sierra(ton)',
                    'costa_norte': 'oferta_costa_norte(ton)',
                    'selva' : 'oferta_selva(ton)',
                    'costa_sur': 'oferta_costa_sur(ton)',
                    'total': 'total(ton)'}, inplace=True)
df3.rename(columns={'ica': 'precio_ica(soles/kg)',
                    'la_libertad': 'precio_la_libertad(soles/kg)',
                    'lima': 'precio_lima(soles/kg)'},inplace=True)
```

```python
print(df1.shape)
print(df2.shape)
print(df3.shape)
```

# Outher join para conservar todos los datos

```python
# Unir todo manteniendo todas las fechas
df_completo = df1.merge(df2, on='fecha', how='outer', suffixes=('_df1', '_df2'))
df_completo = df_completo.merge(df3, on='fecha', how='outer')
```

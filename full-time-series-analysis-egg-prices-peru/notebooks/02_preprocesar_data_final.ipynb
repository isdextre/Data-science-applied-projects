```python
import pandas as pd
```

```python
# leemos los csvs
df1 = pd.read_csv('datos_tabla1.csv')
df2 = pd.read_csv('datos_tabla2.csv')
df3 = pd.read_csv('datos_tabla3.csv')
```

```python
# es necesario para unir, una columna año
# sacar el año del nombre del pdf
df1['anio'] = df1['pdf_name'].str.extract(r'(\d{4})').astype(int)
df2['anio'] = df2['pdf_name'].str.extract(r'(\d{4})').astype(int)
df3['anio'] = df3['pdf_name'].str.extract(r'(\d{4})').astype(int)
```

```python
# creamos la columna fecha
df1['fecha'] = pd.to_datetime(
    dict(year=df1['anio'], month=df1['mes_num'], day=df1['dia'])
)

df2['fecha'] = pd.to_datetime(
    dict(year=df2['anio'], month=df2['mes_num'], day=df2['dia'])
)

df3['fecha'] = pd.to_datetime(
    dict(year=df3['anio'], month=df3['mes_num'], day=df3['dia'])
)
```

```python
# e nu mismo pdf hay varios días por lo que no es un ID
df1 = df1.drop(columns=['pdf_name'])
df2 = df2.drop(columns=['pdf_name'])
df3 = df3.drop(columns=['pdf_name'])
```

# Merge df1 y df2

```python
# debemos crear una sola base para las 3 tablas, el ID será pdf_name
# merge de df1, df2  por pdf name
df_1_2 = df1.merge(df2, on=['dia', 'mes_num', 'anio'], how='inner')
```

```python
# verificamos duplicados
df1.duplicated(subset=['pdf_name', 'mes_num', 'dia']).sum()
```

```python
#revisamos si mes x y mes si hay datos que difieren
df_diff = df_1_2[df_1_2['mes_x'] != df_1_2['mes_y']]
df_diff.head()
```

```python
# no hay datos que diferean entonces ya podemos borrar mes_y
df_1_2.drop(columns=['mes_y'], inplace=True)
# rename as mes
df_1_2.rename(columns={'mes_x': 'mes'}, inplace=True)
```
